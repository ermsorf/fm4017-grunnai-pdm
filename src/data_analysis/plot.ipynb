```python
from pathlib import Path
# import src.config
ROOT = Path(r"C:\projects\fm4017-grunnai-pdm")

import matplotlib.pyplot as plt
import pandas as pd
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score


# ROOT = Path(src.config.BASE_DIR)
OUTPUT_FOLDER = ROOT / "plots" / "report"
OUTPUT_FOLDER.mkdir(parents=True, exist_ok=True)

# Load RNN data
rnn_vibr1 = pd.read_csv(ROOT / "src" / "models" / "rnn" / "vibr1" / "rnn_predictions.csv")
rnn_vibr2 = pd.read_csv(ROOT / "src" / "models" / "rnn" / "vibr2" / "rnn_predictions.csv")
rnn_vibr1["timestamp"] = pd.to_datetime(rnn_vibr1["timestamp"])
rnn_vibr2["timestamp"] = pd.to_datetime(rnn_vibr2["timestamp"])
rnn_vibr1_actual = rnn_vibr1["Scada.GRUN.AGG2.G2.LAGER.M_VIBR1.csv_actual"]
rnn_vibr1_predicted = rnn_vibr1["Scada.GRUN.AGG2.G2.LAGER.M_VIBR1.csv_predicted"]
rnn_vibr2_actual = rnn_vibr2["Scada.GRUN.AGG2.G2.LAGER.M_VIBR2.csv_actual"]
rnn_vibr2_predicted = rnn_vibr2["Scada.GRUN.AGG2.G2.LAGER.M_VIBR2.csv_predicted"]

# Load LSTM data
lstm_vibr1 = pd.read_csv(ROOT / "src" / "models" / "lstm" / "vibr1" / "lstm_predictions.csv")
lstm_vibr2 = pd.read_csv(ROOT / "src" / "models" / "lstm" / "vibr2" / "lstm_predictions.csv")
lstm_vibr1["timestamp"] = pd.to_datetime(lstm_vibr1["timestamp"])
lstm_vibr2["timestamp"] = pd.to_datetime(lstm_vibr2["timestamp"])
lstm_vibr1_actual = lstm_vibr1["Scada.GRUN.AGG2.G2.LAGER.M_VIBR1.csv_actual"]
lstm_vibr1_predicted = lstm_vibr1["Scada.GRUN.AGG2.G2.LAGER.M_VIBR1.csv_predicted"]
lstm_vibr2_actual = lstm_vibr2["Scada.GRUN.AGG2.G2.LAGER.M_VIBR2.csv_actual"]
lstm_vibr2_predicted = lstm_vibr2["Scada.GRUN.AGG2.G2.LAGER.M_VIBR2.csv_predicted"]

# Load GRU data
gru_vibr1 = pd.read_csv(ROOT / "src" / "models" / "gru" / "vibr1" / "gru_predictions.csv")
gru_vibr2 = pd.read_csv(ROOT / "src" / "models" / "gru" / "vibr2" / "gru_predictions.csv")
gru_vibr1["timestamp"] = pd.to_datetime(gru_vibr1["timestamp"])
gru_vibr2["timestamp"] = pd.to_datetime(gru_vibr2["timestamp"])
gru_vibr1_actual = gru_vibr1["Scada.GRUN.AGG2.G2.LAGER.M_VIBR1.csv_actual"]
gru_vibr1_predicted = gru_vibr1["Scada.GRUN.AGG2.G2.LAGER.M_VIBR1.csv_predicted"]
gru_vibr2_actual = gru_vibr2["Scada.GRUN.AGG2.G2.LAGER.M_VIBR2.csv_actual"]
gru_vibr2_predicted = gru_vibr2["Scada.GRUN.AGG2.G2.LAGER.M_VIBR2.csv_predicted"]

# Additional LSTM data without needle 1
lstm_vibr1_nal1 = pd.read_csv(ROOT / "src" / "models" / "lstm" / "vibr1_nal1" / "lstm_predictions.csv")
lstm_vibr2_nal1 = pd.read_csv(ROOT / "src" / "models" / "lstm" / "vibr2_nal1" / "lstm_predictions.csv")
lstm_vibr1_nal1["timestamp"] = pd.to_datetime(lstm_vibr1_nal1["timestamp"])
lstm_vibr2_nal1["timestamp"] = pd.to_datetime(lstm_vibr2_nal1["timestamp"])
lstm_vibr1_nal1_actual = lstm_vibr1_nal1["Scada.GRUN.AGG2.G2.LAGER.M_VIBR1.csv_actual"]
lstm_vibr1_nal1_predicted = lstm_vibr1_nal1["Scada.GRUN.AGG2.G2.LAGER.M_VIBR1.csv_predicted"]
lstm_vibr2_nal1_actual = lstm_vibr2_nal1["Scada.GRUN.AGG2.G2.LAGER.M_VIBR2.csv_actual"]
lstm_vibr2_nal1_predicted = lstm_vibr2_nal1["Scada.GRUN.AGG2.G2.LAGER.M_VIBR2.csv_predicted"]
```

```python
# Calculate and print metrics
print("rnn_vibr1")
print("MAE:", mean_absolute_error(rnn_vibr1_actual, rnn_vibr1_predicted))
print("RMSE:", mean_squared_error(rnn_vibr1_actual, rnn_vibr1_predicted) ** 0.5)
print("R2:", r2_score(rnn_vibr1_actual, rnn_vibr1_predicted))

print("\nrnn_vibr2")
print("MAE:", mean_absolute_error(rnn_vibr2_actual, rnn_vibr2_predicted))
print("RMSE:", mean_squared_error(rnn_vibr2_actual, rnn_vibr2_predicted) ** 0.5)
print("R2:", r2_score(rnn_vibr2_actual, rnn_vibr2_predicted))

print("\nlstm_vibr1")
print("MAE:", mean_absolute_error(lstm_vibr1_actual, lstm_vibr1_predicted))
print("RMSE:", mean_squared_error(lstm_vibr1_actual, lstm_vibr1_predicted) ** 0.5)
print("R2:", r2_score(lstm_vibr1_actual, lstm_vibr1_predicted))

print("\nlstm_vibr2")
print("MAE:", mean_absolute_error(lstm_vibr2_actual, lstm_vibr2_predicted))
print("RMSE:", mean_squared_error(lstm_vibr2_actual, lstm_vibr2_predicted) ** 0.5)
print("R2:", r2_score(lstm_vibr2_actual, lstm_vibr2_predicted))

print("\ngru_vibr1")
print("MAE:", mean_absolute_error(gru_vibr1_actual, gru_vibr1_predicted))
print("RMSE:", mean_squared_error(gru_vibr1_actual, gru_vibr1_predicted) ** 0.5)
print("R2:", r2_score(gru_vibr1_actual, gru_vibr1_predicted))

print("\ngru_vibr2")
print("MAE:", mean_absolute_error(gru_vibr2_actual, gru_vibr2_predicted))
print("RMSE:", mean_squared_error(gru_vibr2_actual, gru_vibr2_predicted) ** 0.5)
print("R2:", r2_score(gru_vibr2_actual, gru_vibr2_predicted))
```
